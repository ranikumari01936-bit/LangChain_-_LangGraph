```python
import os
from groq import Groq
from dotenv import load_dotenv

load_dotenv()
```

```python
GROQ_API_KEY = os.getenv("GROQ_API_KEY")
if not GROQ_API_KEY:
    raise ValueError("Set GROQ_API_KEY in your environment before running this cell")
```

```python
client = Groq(api_key=GROQ_API_KEY)
models = client.models.list()
for x in models.data:
    print(x.id)
```

```python
responce = client.chat.completions.create(model="llama-3.3-70b-versatile", 
messages=[{"role": "user", "content": "Translate good evening into hindi"}])
print(responce.choices[0].message.content)
```
